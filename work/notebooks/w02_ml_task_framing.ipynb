## 1. My lane as an ML task (type)

Ranking / scoring (primary) + signal analysis via segmentation (secondary).

Primary task: rank pages by the size of their search-performance gap relative to ranking position, so content writers and SEO strategists know which pages to fix first.

Secondary task: segment pages by position band, compare high-CTR vs low-CTR pages inside each band, and analyze which ranking signals (age, freshness, word count, content type, and other quality indicators) differ — to explain why the gap exists.

Why this type: the core decision is “which pages first?” (ranking/scoring). Segmentation and signal comparison explain the score and turn it into actionable recommendations.

## 2. Target or proxy

Target: a page-level search-performance gap score (priority score).

How the gap is defined (observed from data, not a product rule):
1. Segment pages by position band (e.g. 1–3, 4–10, 11–20).
2. Inside each band, compare ctr to the band’s typical ctr (median or mean).
3. A page has a gap when:
   - avg_position is in a visible band (focus: 1–10)
   - impressions meet a minimum volume (e.g. ≥ 100)
   - ctr is substantially below the band typical (e.g. ≥ 30% below median)

The ranking target is a continuous score from the size of that gap
(how far below the band typical the page sits, optionally weighted by impressions).
Higher score = higher priority to review first.

Label source: measured Search Console fields (avg_position, impressions, ctr) — observed outcome, not a hand-written FlyRank flag.

## 3. Success metric

Primary metric: Precision@K with K = 50.

Meaning: of the top 50 pages ranked by our gap score, what fraction actually meet the gap definition (low ctr for their position band, with enough impressions).

Why this metric:
- Matches the decision (“which ones first?”)
- Can be computed on a simple baseline today (e.g. sort by lowest ctr only)
- Lets us compare: baseline Precision@50 vs our scored ranking

Secondary (for segmentation / signal analysis): grouped comparisons and effect sizes
(e.g. median word_count or content age on high-gap vs low-gap pages within the same band).
These are directional only.

“Good” = Precision@50 clearly higher than a transparent baseline rule.

## 4. The unit of analysis, as a real dataframe

One row = one content item (one pseudonymized page).

Grain: content_id in the starter dataset (content_hash_id in the warehouse).
Metrics are trailing-window aggregates on the starter CSV.
Position-band segmentation and gap flags are derived at this same grain
(one band label and one gap score per page).

IDs (content_id, client_id) are for joining, grouping, and train/test splits only — never features.
